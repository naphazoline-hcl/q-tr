# 提出ファイルについて

作成した戦略は、**ZIPファイル** にまとめて提出してください。ZIPファイルには、以下の2点を必ず含めます。

- `submission.py`：評価時に実行される、売買戦略を実装したPythonファイル
- 戦略レポート：戦略の考え方、使用した指標、売買ルール、検証結果などを説明するレポート

戦略レポートのファイル名と形式は、別途指定された提出要項に従ってください。

このNotebookでは、下のセルにある `NAME` を提出する戦略フォルダ名に変更して実行すると、提出用のZIPファイルを作成できます。対象フォルダに `submission.py` がない場合は、ZIPを作成せずエラーで停止します。

```python
from pathlib import Path
import zipfile

NAME = "sample01_short_reversal"  # ZIP化する戦略フォルダ名に変更

strategies_dir = Path.cwd()
if strategies_dir.name != "strategies":
    strategies_dir = Path("strategies")

strategy_dir = strategies_dir / NAME
if not strategy_dir.is_dir():
    raise FileNotFoundError(f"戦略フォルダが見つかりません: {strategy_dir.resolve()}")

submission_file = strategy_dir / "submission.py"
if not submission_file.is_file():
    raise FileNotFoundError(
        f"提出条件を満たしていません。submission.py が見つかりません: {submission_file.resolve()}"
    )

zip_path = strategies_dir / f"{NAME}.zip"
junk_dirs = {"__pycache__", ".ipynb_checkpoints", ".git", ".venv", "__MACOSX"}

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as archive:
    for path in sorted(strategy_dir.rglob("*")):
        relative = path.relative_to(strategy_dir)
        if (
            not path.is_file()
            or any(part in junk_dirs for part in relative.parts)
            or path.name == ".DS_Store"
            or path.name.startswith("._")
        ):
            continue
        archive.write(path, Path(NAME) / relative)

print("提出ZIP:", zip_path.resolve())
with zipfile.ZipFile(zip_path) as archive:
    print("ZIP内容:", archive.namelist())
```
